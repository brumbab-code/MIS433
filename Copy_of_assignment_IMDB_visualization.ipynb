```python
# Select all movies released after (>=) 2010 and with IMDB_Rating>=8.5
# Show their Series_Title, Released_Year, Certificate, and Gross.
# Sort them in descending order of "Gross"

(
    imdb
    .sort_values('Gross', ascending=False)
    .query('Released_Year >= 2010 and IMDB_Rating>=8.5')
    [['Series_Title', 'Released_Year', 'Certificate','Gross']]
)
```

```python
# Does the sorting result look right to you? What's the problem?

##Gisaengchung is not the highest grossed movie, but it is sorted as the highest.
#The problem is the data type for gross is identified as an object when it should be identified as a float. Since it is text, Panas sorts it by character, so 53,367,844 comes before 335,451,311, since 5 comes after 3.
```

```python
# Resolve this problem of "Gross" and convert its data type to float
# Hint: You may find this webpage useful:
# https://stackoverflow.com/questions/28986489/how-to-replace-text-in-a-column-of-a-pandas-dataframe


imdb['Gross']=imdb['Gross'].str.replace(',','').astype(float)
imdb.Gross

```

```python
# Next, redo the sorting on Gross

# Select all movies released after (>=) 2010 and with IMDB_Rating>=8.5
# Show their Series_Title, Released_Year, Certificate, and Gross.
# Sort them in descending order of "Gross"

(
    imdb
    .sort_values('Gross', ascending=False)
    .query('Released_Year >= 2010 and IMDB_Rating>=8.5')
    [['Series_Title', 'Released_Year', 'Certificate','Gross']]
)

```

```python
# Add a new column "Runtime_min" by removing the substring " min" in "Runtime"
# Set its data type as int
# Hint: https://stackoverflow.com/questions/36505847/substring-of-an-entire-column-in-pandas-dataframe

imdb['Runtime_min'] = imdb['Runtime'].str[:-4].astype(int)
imdb['Runtime_min']
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Create a scatterplot of the two scores, "IMDB_Rating" and "Meta_score".
# In a sentence or two, describe what this pair of scores suggests about the movies.

sns.scatterplot(data = imdb, x = 'IMDB_Rating', y = 'Meta_score')
plt.show()

#Movies with a higher IMDB_Rating tend to get a higher Meta score but the scores are widley spread. It seems that meta score and imdb ratings disagree often, and they agree most on the very highest rated movies.
```

```python
# Fit a trendline to show the relationship between the two scores.
# Hint: sns.regplot()
# Try a few values of the `order` argument (polynomial degree) for the trendline.

sns.regplot( data = imdb, x = 'IMDB_Rating', y = 'Meta_score', order = 3)
plt.show()
```

```python
# Do any data points in the scatterplot surprise you?
# Try to identify a few such movies.
(
    imdb
    .query('Meta_score < 40')
    [['Series_Title','IMDB_Rating','Meta_score']]
)


# Yes, a few movies got below an 8 IMDB Rating and below 40 Meta score which is extremely low. Tropa de Elite got an imdb rating of 8, but got a meta score of 33, which suggests that the audience enjoyed the movie much more than the critics did.
# It was also suprising that the highest IMDB rated movie (The Shawshank Redemption) was only rated a meta score of 80.
```
